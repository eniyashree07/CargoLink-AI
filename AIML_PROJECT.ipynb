### Interactive Prediction Widget

Let's create an interactive widget using `ipywidgets` to input new applicant details and get real-time predictions from our trained Logistic Regression model. This will allow for easy experimentation with different scenarios.

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import pandas as pd

# Create input widgets for each feature
applicant_income_widget = widgets.IntText(
    value=5000,
    description='Applicant Income:',
    disabled=False
)

coapplicant_income_widget = widgets.IntText(
    value=2000,
    description='Coapplicant Income:',
    disabled=False
)

loan_amount_widget = widgets.IntText(
    value=250000,
    description='Loan Amount:',
    disabled=False
)

loan_amount_term_widget = widgets.IntText(
    value=360,
    description='Loan Amount Term (months):',
    disabled=False
)

credit_history_widget = widgets.FloatText(
    value=1.0,
    description='Credit History (0 or 1):',
    disabled=False
)

# Create a button to trigger prediction
predict_button = widgets.Button(
    description='Predict Loan Status',
    disabled=False,
    button_style='success', # 'success', 'info', 'warning', 'danger' or ''
    tooltip='Click to predict',
    icon='check'
)

# Output widget to display results
output_area = widgets.Output()

def on_predict_button_clicked(b):
    with output_area:
        clear_output()
        # Get values from widgets
        new_data = {
            'Applicant_Income': [applicant_income_widget.value],
            'Coapplicant_Income': [coapplicant_income_widget.value],
            'Loan_Amount': [loan_amount_widget.value],
            'Loan_Amount_Term': [loan_amount_term_widget.value],
            'Credit_History': [credit_history_widget.value]
        }

        new_df = pd.DataFrame(new_data)

        print("--- Input Data ---")
        display(new_df)

        # Scale the new data using the fitted scaler
        # Ensure scaler and log_reg are available from previous steps
        if 'scaler' not in globals() or 'log_reg' not in globals():
            print("Error: Scaler or Logistic Regression model not found. Please run previous cells.")
            return

        new_scaled_data = scaler.transform(new_df)

        # Make prediction
        prediction = log_reg.predict(new_scaled_data)
        prediction_proba = log_reg.predict_proba(new_scaled_data)

        loan_status = 'Approved' if prediction[0] == 1 else 'Rejected'
        probability_approved = prediction_proba[0][1] # Probability of being approved (class 1)

        print("\n--- Prediction Result ---")
        print(f"Predicted Loan Status: {loan_status}")
        print(f"Probability of Approval: {probability_approved:.2f}")

# Attach the button click event to the function
predict_button.on_click(on_predict_button_clicked)

# Display the widgets
print("Enter new applicant details and click 'Predict Loan Status':")
display(
    applicant_income_widget,
    coapplicant_income_widget,
    loan_amount_widget,
    loan_amount_term_widget,
    credit_history_widget,
    predict_button,
    output_area
)
```
